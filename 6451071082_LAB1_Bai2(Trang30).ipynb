```python


import pandas as pd
import matplotlib.pyplot as plt
import io
import requests

print("Lê Quốc Trung - 6451071082")
try:
    df = pd.read_excel('tb_tracking.xlsx', header=None)
except Exception:
    file_id = '1Xizh3zvtMF4IdRNlueb74meOXk80-CHc'
    download_url = f'https://drive.google.com/uc?export=download&id={file_id}'
    response = requests.get(download_url)
    df = pd.read_excel(io.BytesIO(response.content), header=None)

df.columns = ['id', 'device_id', 'co_level', 'humidity', 'temperature', 'time']
df['time'] = pd.to_datetime(df['time'])
df_filtered = df[df['time'] >= '2018-01-03'].copy()
df_filtered['minutes'] = df_filtered['time'].dt.hour * 60 + df_filtered['time'].dt.minute

fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(10, 12))

axes[0].scatter(df_filtered['minutes'], df_filtered['co_level'], alpha=0.5, color='red', s=15)
axes[0].set_title('Scatter Plot: Nong do khi CO (co_level) theo Minutes')
axes[0].set_xlabel('Thoi diem trong ngay (Phut)')
axes[0].set_ylabel('Nong do CO')
axes[0].grid(True, linestyle='--', alpha=0.6)

axes[1].scatter(df_filtered['minutes'], df_filtered['humidity'], alpha=0.5, color='blue', s=15)
axes[1].set_title('Scatter Plot: Do am (humidity) theo Minutes')
axes[1].set_xlabel('Thoi diem trong ngay (Phut)')
axes[1].set_ylabel('Do am (%)')
axes[1].grid(True, linestyle='--', alpha=0.6)

axes[2].scatter(df_filtered['minutes'], df_filtered['temperature'], alpha=0.5, color='green', s=15)
axes[2].set_title('Scatter Plot: Nhiet do (temperature) theo Minutes')
axes[2].set_xlabel('Thoi diem trong ngay (Phut)')
axes[2].set_ylabel('Nhiet do (Do C)')
axes[2].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
